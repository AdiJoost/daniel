# Generates the dataset to use.

```python
from initNotebook import initNotebook
initNotebook()
import yfinance as yf
from config.rootPath import getRootPath
import pandas as pd
```

```python
from_date = "2015-01-01"  # YYYY-MM-DD
to_date = "2019-12-31"    # YYYY-MM-DD
tickers = ["SLV", "IAU", "SIL", "GDX", "GDXJ", "UUP", "FXE", "FXY", "FXB", "FXA", "FXC", "CEW"]

tickers.append("^GSPC") # The standart against which the agent should compete (^GSPC is the S&P 500)

data = yf.download(
    tickers,
    start=from_date,
    end=to_date
)
```

```python
data.head()
```

```python
data = data.copy()
data.index.name = 'date'

# stack tickers into a long DataFrame: columns -> ['date','ticker','Open','High',...]
df = data.stack(level=1).reset_index()

# normalize column names and make date a datetime dtype
df['date'] = pd.to_datetime(df['date'])
df.columns = [c.lower() if isinstance(c, str) else c for c in df.columns]

# keep only core columns
df = df[['date', 'ticker', 'open', 'high', 'low', 'close', 'volume']]
```

```python
df.head()
```

```python
out_dir = getRootPath().joinpath('data', 'yfinance')
out_dir.mkdir(parents=True, exist_ok=True)
out_file = out_dir.joinpath(f'metals_tickers_{from_date}_to_{to_date}_tidy.csv')
df.to_csv(out_file, index=False)
print(f'Saved CSV to: {out_file} (rows={len(df)})')
```
